**Step 1: Understand the Data Structure**

If you have ONE customer, how would you store it?

customer = {
    "name":"Alice",
    "country":"Canada",
    "subscription":"Premium",
    "hours":25
}

**This is one dictionary.**

**What if Netflix has 1000 customers?**

One dictionary isn't enough.We need something that can hold MANY dictionaries.
That's where a list comes in.

customers = [ {customer1},{customer2},{customer3},{customer4}

The list stores many customers.Each customer is stored as a dictionary.

#List of Dicitionary

```python
L=[]
while True:
  D={}
  Cname=input("Enter name:")
  D['name']=Cname
  CCountry=input("Enter Country:")
  D['country']=CCountry
  CSub=input("Enter Subscription:")
  D['Sub']=CSub
  hours=int(input("Enter hours:"))
  D['hours']=hours
  print("First Dicitionary of Customer created", D)
  L.append(D)  #append individual dicitionary to global List

  ch=input("Add more? y/n:")
  if ch.lower()=='n':
    break

print("The whole database of customers in the list",L)
```

```python
customers = L

while True:

    print("\n==============================")
    print(" Netflix Analytics Dashboard ")
    print("==============================")
    print()
    print("1. Add Customer")
    print("2. View All Customers")
    print("3. Find Heavy Watchers")
    print("4. Find Top Viewer")
    print("5. Calculate Average Watch Hours")
    print("6. Subscription Report")
    print("7. Upgrade Subscription")
    print("8. Search Customer by Country")
    print("9. Show Unique Countries and Genres")
    print("10. Exit")
    print()
    choice = input("\nEnter your choice: ")
    print (choice)
    # 1. Add Customer
    if choice == "1":
        name = input("Enter customer name: ")
        country = input("Enter country: ")
        Sub = input("Enter subscription (Basic/Standard/Premium): ")
        hours = (input("Enter watch hours: "))

        customer = {
            "name": name,
            "country": country,
            "Sub": Sub,
            "hours": hours
        }

        customers.append(customer)

        print("Customer added successfully!")

    # 2. View All Customers
    elif choice == "2":

        if len(customers) == 0:
            print("No customers found.")

        else:
            print("\n===== All Customers =====")

            for customer in customers:
                print(customer)


    # 3. Find Heavy Watchers
    elif choice == "3":

        print("\n===== Heavy Watchers (More Than 20 Hours) =====")

        found = False

        for customer in customers:

            if customer["hours"] > 20:
                print(customer)
                found = True

        if found == False:
            print("No heavy watchers found.")


    # 4. Find Top Viewer
    elif choice == "4":

        if len(customers) == 0:
            print("No customers found.")

        else:
            top_viewer = customers[0]

            for customer in customers:

                if customer["hours"] > top_viewer["hours"]:
                    top_viewer = customer

            print("\n===== Top Viewer =====")
            print("Name:", top_viewer["name"])
            print("Watch Hours:", top_viewer["hours"])


    # 5. Calculate Average Watch Hours
    elif choice == "5":

        if len(customers) == 0:
            print("No customers found.")

        else:
            total_hours = 0

            for customer in customers:
                total_hours = total_hours + customer["hours"]

            average_hours = total_hours / len(customers)

            print("\nAverage Watch Hours:", average_hours)


    # 6. Subscription Report
    elif choice == "6":

        subscription_count = {}

        for customer in customers:

            subscription = customer["Sub"]

            if subscription in subscription_count:
                subscription_count[subscription] = subscription_count[subscription] + 1

            else:
                subscription_count[subscription] = 1

        print("\n===== Subscription Report =====")

        for subscription, count in subscription_count.items():
            print(subscription, ":", count)


    # 7. Upgrade Subscription
    elif choice == "7":

        name = input("Enter customer name to upgrade: ")

        found = False

        for customer in customers:

            if customer["name"].lower() == name.lower():

                current_subscription = customer["Sub"]

                if current_subscription == "Basic":
                    customer["subscription"] = "Standard"

                elif current_subscription == "Standard":
                    customer["subscription"] = "Premium"

                elif current_subscription == "Premium":
                    print("Customer already has Premium subscription.")
                    found = True
                    break

                else:
                    print("Invalid subscription level.")
                    found = True
                    break

                print(
                    "Subscription upgraded from",
                    current_subscription,
                    "to",
                    customer["subscription"]
                )

                found = True
                break

        if found == False:
            print("Customer not found.")


    # 8. Search Customer by Country
    elif choice == "8":

        country = input("Enter country: ")

        found = False

        print("\n===== Customers from", country, "=====")

        for customer in customers:

            if customer["country"].lower() == country.lower():
                print(customer[name] + ",")
                found = True

        if found == False:
            print("No customers found from", country)


    # 9. Exit Program
    elif choice == "9":

        print("Thank you for using Netflix Customer Analytics System!")
        break


    # Invalid Choice
    else:

        print("Invalid choice. Please try again.")




```
